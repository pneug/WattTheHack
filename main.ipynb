```python
import torch
import cv2 as cv

assert torch.__version__
assert cv.__version__
assert hasattr(torch, 'cuda')
assert hasattr(cv, 'cuda') or True

print(f"torch: {torch.__version__}")
print(f"torch cuda build: {torch.version.cuda}")
print(f"torch cuda available: {torch.cuda.is_available()}")
print(f"torch cuda device count: {torch.cuda.device_count()}")
print(f"cv2: {cv.__version__}")
print(f"cv2 cuda devices: {cv.cuda.getCudaEnabledDeviceCount() if hasattr(cv, 'cuda') else 0}")

if not torch.cuda.is_available() or torch.cuda.device_count() == 0:
    print("Warning: CUDA device is not available in this environment. Install NVIDIA drivers and a CUDA-capable GPU to use GPU acceleration.")
```
